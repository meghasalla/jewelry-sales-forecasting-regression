# Jewelry Sales: Business and Data Understanding

**Objective:** understand the synthetic store-month dataset and connect each analysis to a business decision.

> All data is synthetic and must not be represented as real retailer data.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
data_path = project_root / 'data' / 'raw' / 'jewelry_sales_monthly.csv'
data = pd.read_csv(data_path, parse_dates=['month'])
data.shape, data.head()
```

## Data quality checks

```python
print(data.info())
display(data.isna().sum().sort_values(ascending=False).head())
display(data.describe(include='all').T)
```

## Monthly sales trend

```python
monthly = data.groupby('month', as_index=False)['sales_lakh'].sum()
plt.figure(figsize=(12, 4))
sns.lineplot(data=monthly, x='month', y='sales_lakh', marker='o')
plt.title('Total Synthetic Jewelry Sales by Month')
plt.ylabel('Sales (lakh INR)')
plt.show()
```

## Festival comparison

```python
display(data.groupby('festival_month')['sales_lakh'].agg(['count','mean','median','std']).round(2))
sns.boxplot(data=data, x='festival_month', y='sales_lakh')
plt.title('Sales Distribution: Regular vs Festival Months')
plt.show()
```
