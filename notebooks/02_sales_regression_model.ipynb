# Jewelry Sales Regression Model

Build an interpretable baseline, evaluate it on future months, and translate the result into business recommendations.

```python
from pathlib import Path
import sys
import pandas as pd

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(project_root / 'src'))
from train_model import FEATURES, TARGET, build_model, load_data
data = load_data()
data.shape
```

## Time-based split

The latest six months are held out so the model is tested on future periods rather than a random sample of the past.

```python
test_months = sorted(data['month'].unique())[-6:]
train = data[~data['month'].isin(test_months)]
test = data[data['month'].isin(test_months)]
train['month'].min(), train['month'].max(), test['month'].min(), test['month'].max()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
model = build_model()
model.fit(train[FEATURES], train[TARGET])
predictions = model.predict(test[FEATURES])
metrics = {
    'MAE': mean_absolute_error(test[TARGET], predictions),
    'RMSE': mean_squared_error(test[TARGET], predictions) ** 0.5,
    'R2': r2_score(test[TARGET], predictions),
}
pd.Series(metrics).round(3)
```

## Interpretation prompts

1. Is the MAE small enough for the planning decision?
2. Does the model beat a mean or previous-month baseline?
3. Are errors larger for specific stores or festival months?
4. Which inputs are controllable by management?
5. What pilot would validate business impact without overclaiming causality?
